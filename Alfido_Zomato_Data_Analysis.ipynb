```python
import matplotlib.pyplot as plt

top_cuisines = df['Cuisine'].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_cuisines.plot(kind='bar')
plt.title('Top 10 Most Common Cuisines')
plt.xlabel('Cuisine')
plt.ylabel('Number of Records')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df['Average_Rating'], bins=20)

plt.title('Distribution of Average Restaurant Ratings')
plt.xlabel('Average Rating')
plt.ylabel('Number of Records')

plt.tight_layout()
plt.show()
```

```python
top_cities = df['City'].value_counts().head(10)

plt.figure(figsize=(10, 6))
top_cities.plot(kind='bar')
plt.title('Top 10 Cities by Number of Records')
plt.xlabel('City')
plt.ylabel('Number of Records')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(df['Prices'], df['Average_Rating'], alpha=0.3)

plt.title('Price vs Average Restaurant Rating')
plt.xlabel('Price')
plt.ylabel('Average Rating')

plt.tight_layout()
plt.show()
```

```python
print("Top 10 Cuisines:")
print(df['Cuisine'].value_counts().head(10))

print("\nTop 10 Cities:")
print(df['City'].value_counts().head(10))

print("\nAverage Rating:")
print("Minimum:", df['Average_Rating'].min())
print("Maximum:", df['Average_Rating'].max())
print("Mean:", df['Average_Rating'].mean())

print("\nPrice:")
print("Minimum:", df['Prices'].min())
print("Maximum:", df['Prices'].max())
print("Mean:", df['Prices'].mean())
```

# Zomato Dataset Analysis

## Project Overview

This project was completed as part of the Alfido Tech Data Science Internship.

The objective of this project is to perform Exploratory Data Analysis (EDA) on a Zomato restaurant dataset and identify useful patterns related to cuisines, cities, restaurant ratings, and prices.

## Dataset

The dataset contains information about restaurants, cuisines, cities, ratings, votes, and prices.

- Number of records: 123,657
- Number of columns: 26
- Missing values: None
- Dataset source: Kaggle
- Dataset name: Zomato Restaurant Dataset

## Tools and Technologies

- Python
- Pandas
- Matplotlib
- Google Colab

## Analysis Performed

The following analysis was performed:

1. Loaded and inspected the dataset.
2. Checked the dataset shape and data types.
3. Checked for missing values.
4. Generated summary statistics.
5. Identified the most common cuisines.
6. Identified cities with the highest number of records.
7. Analyzed the distribution of restaurant ratings.
8. Examined the relationship between price and average rating.

## Key Insights

- Beverages is the most frequently represented cuisine/category with 39,441 records.
- Hyderabad has the highest number of records with 15,613.
- Restaurant ratings range from 3.0 to 4.65, with an average rating of approximately 3.89.
- The average listed price is approximately ₹241.38.
- Prices range from approximately ₹0.95 to ₹12,024.

## Conclusion

The analysis provides an overview of restaurant data across different cuisines and cities. It also helps understand restaurant ratings and pricing patterns. The visualizations make it easier to identify important trends in the dataset.
