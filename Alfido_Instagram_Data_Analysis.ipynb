```python
import pandas as pd

file_path = "/content/instagram_data/posts.csv"

df = pd.read_csv(file_path)

df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nSummary Statistics:")
print(df.describe())
```

```python
import matplotlib.pyplot as plt

post_types = df['Post_Type'].value_counts()

post_types.plot(kind='bar')

plt.title('Distribution of Instagram Post Types')
plt.xlabel('Post Type')
plt.ylabel('Number of Posts')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
likes_by_type = df.groupby('Post_Type')['likes'].mean()

likes_by_type.plot(kind='bar')

plt.title('Average Likes by Post Type')
plt.xlabel('Post Type')
plt.ylabel('Average Likes')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
likes_by_type = df.groupby('Post_Type')['likes'].mean()

likes_by_type.plot(kind='bar')

plt.title('Average Likes by Post Type')
plt.xlabel('Post Type')
plt.ylabel('Average Likes')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
comments_by_type = df.groupby('Post_Type')['comments'].mean()

comments_by_type.plot(kind='bar')

plt.title('Average Comments by Post Type')
plt.xlabel('Post Type')
plt.ylabel('Average Comments')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
print("KEY INSIGHTS")
print("-" * 40)

print(f"Total posts: {len(df)}")
print(f"Most common post type: {df['Post_Type'].mode()[0]}")
print(f"Average likes: {df['likes'].mean():,.2f}")
print(f"Average comments: {df['comments'].mean():,.2f}")
print(f"Maximum likes: {df['likes'].max():,}")
print(f"Maximum comments: {df['comments'].max():,}")

print("\nAverage likes by post type:")
print(df.groupby('Post_Type')['likes'].mean().sort_values(ascending=False))

print("\nAverage comments by post type:")
print(df.groupby('Post_Type')['comments'].mean().sort_values(ascending=False))
```
