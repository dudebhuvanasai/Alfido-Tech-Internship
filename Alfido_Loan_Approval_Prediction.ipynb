```python
import matplotlib.pyplot as plt

df['loan_status'].value_counts().plot(kind='bar')

plt.title('Loan Approval Status Distribution')
plt.xlabel('Loan Status')
plt.ylabel('Number of Applicants')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.hist(df['cibil_score'], bins=20)

plt.title('Distribution of CIBIL Scores')
plt.xlabel('CIBIL Score')
plt.ylabel('Number of Applicants')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

plt.scatter(df['income_annum'], df['loan_amount'], alpha=0.3)

plt.title('Annual Income vs Loan Amount')
plt.xlabel('Annual Income')
plt.ylabel('Loan Amount')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

df.boxplot(column='cibil_score', by='loan_status')

plt.title('CIBIL Score by Loan Approval Status')
plt.suptitle('')
plt.xlabel('Loan Status')
plt.ylabel('CIBIL Score')

plt.tight_layout()
plt.show()
```
